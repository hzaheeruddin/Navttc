Some rows have invalid `purpose` values (a scraping error produced things like addresses instead of "For Rent"/"For Sale"). Select all rows where `purpose` is **neither** `"For Rent"` nor `"For Sale"`, and store them in a variable (e.g. `wrong_scrap`).

### Prerequisites
Runs the setup and the earlier steps this task depends on (output hidden).

```python
%%capture
import numpy as np
import pandas as pd
import re

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

df = pd.read_csv("zameenData.csv", index_col=0)
df = df.rename(columns={"pupose": "purpose"})   # fix the typo in the column name
df.head()

df = df.drop_duplicates()

df = df.drop(columns=["agent", "name"])
```

### Solution

```python
wrong_scrap = df[~df["purpose"].isin(["For Rent", "For Sale"])]
wrong_scrap
```
