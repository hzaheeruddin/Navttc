Based on the spread (std) per land type from the previous task, which land types are the most scattered/variable in price? What might that imply for modeling (e.g. would separate models, or a different target like price-per-marla, make sense for those types)?

### Prerequisites
Runs the setup and the earlier steps this task depends on (output hidden).

```python
%%capture
import numpy as np
import pandas as pd
import re
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

df = pd.read_csv("dataCleansing1_complete.csv", index_col=0)
df.head()

totals = df.groupby("land_type")[["price_cr", "area_value"]].sum()
price_per_marla = (totals["price_cr"] / totals["area_value"]).sort_values()
price_per_marla.plot(kind="bar", figsize=(8, 4), title="Price per marla (Crore) by land type")
plt.ylabel("Crore / marla")
plt.show()

residential = ["House", "Flat", "Residential Plot"]
commercial = ["Building", "Shop", "Office", "Commercial Plot"]
agricultural = ["Plot File", "Farm House", "Agricultural Land"]
industrial = ["Factory", "Industrial Land"]

df1 = df.drop(columns=["location", "price", "area", "added_days", "purpose"])
df1.head()

df1 = df1.rename(columns={"locat": "location"})
df1.columns.tolist()

residential_df = df1[df1["land_type"].isin(residential)].copy()
commercial_df = df1[df1["land_type"].isin(commercial)].copy()
agricultural_df = df1[df1["land_type"].isin(agricultural)].copy()
industrial_df = df1[df1["land_type"].isin(industrial)].copy()
print(len(residential_df), len(commercial_df), len(agricultural_df), len(industrial_df))

loc_counts = residential_df["location"].value_counts()
rare_locations = loc_counts[loc_counts < 40].index
print("Rare locations grouped into 'Others':", len(rare_locations))
residential_df.loc[residential_df["location"].isin(rare_locations), "location"] = "Others"
residential_df["location"].nunique()

location_means = residential_df.groupby("location")["price_cr"].mean()
residential_df["location_encoded"] = residential_df["location"].map(location_means)

stats = residential_df["price_cr"].describe()
print(stats)
print("\nmean:", round(stats["mean"], 3), "| median:", round(stats["50%"], 3))
# mean >> median -> a few very expensive listings pull the mean up (right-skewed distribution).

df_50 = residential_df[residential_df["price_cr"] > 50].copy()
df_50["price_per_marla"] = df_50["price_cr"] / df_50["area_value"]
df_50.head()

plt.figure(figsize=(10, 5))
sns.boxplot(x="land_type", y="price_cr", data=residential_df)
plt.xticks(rotation=90)
plt.yscale("log")
plt.title("price_cr by land type (log scale)")
plt.xlabel("land_type")
plt.ylabel("price_cr (Crore, log scale)")
plt.show()
```

### Answer
The land types with the **largest standard deviation** (typically commercial/large property types such as Building, Farm House or Agricultural Land) are the most scattered in price, because they vary hugely in size. This suggests modelling those types separately or predicting **price per marla** (or `log_price`) instead of raw price, so that size differences do not dominate the error.
