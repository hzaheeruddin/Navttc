## Task 8: Basic Inheritance
Create a parent class `Vehicle` with `brand` and `speed` attributes and a method `describe()` returning a sentence about the vehicle. Create a child class `ElectricVehicle` that inherits from `Vehicle`, adds a `battery_range` attribute, and overrides `describe()` to also mention the battery range (using `super()` to reuse the parent's description).

### Solution

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def describe(self):
        return f"{self.brand} can travel at {self.speed} km/h."


class ElectricVehicle(Vehicle):
    def __init__(self, brand, speed, battery_range):
        super().__init__(brand, speed)
        self.battery_range = battery_range

    def describe(self):
        base = super().describe()
        return f"{base} It has a battery range of {self.battery_range} km."


ev = ElectricVehicle("Tesla", 250, 500)
print(ev.describe())
```
