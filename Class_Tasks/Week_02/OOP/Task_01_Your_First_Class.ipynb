## Task 1: Your First Class
Create a class `Book` with an `__init__` method that accepts `title` and `author`, and stores them as instance attributes. Create one `Book` object and print both attributes.

### Solution

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

book1 = Book("Deep Work", "Cal Newport")
print(book1.title)
print(book1.author)
```
