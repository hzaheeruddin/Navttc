## Task 2: A Method That Returns a Value
Add a method `summary()` to a class `Movie` that has `title` and `year` attributes. `summary()` should return a string like `"Inception (2010)"`. Create an object and print the result of calling `summary()`.

### Solution

```python
class Movie:
    def __init__(self, title, year):
        self.title = title
        self.year = year

    def summary(self):
        return f"{self.title} ({self.year})"

movie1 = Movie("Inception", 2010)
print(movie1.summary())
```
