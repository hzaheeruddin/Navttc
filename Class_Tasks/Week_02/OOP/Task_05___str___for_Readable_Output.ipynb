## Task 5: `__str__` for Readable Output
Add a `__str__` method to the `Employee` class from Task 4 so that `print(employee)` shows something like `"Employee: Ali, Salary: 50000"`. Test it on one object.

### Solution

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"Employee: {self.name}, Salary: {self.salary}"

employee1 = Employee("Ali", 50000)
print(employee1)
```
