## Task 3: Class Attribute vs Instance Attribute
Create a class `Car` with a class attribute `wheels = 4` (shared by all cars) and an instance attribute `brand` (unique per car, set in `__init__`). Create two `Car` objects with different brands and print each one's `wheels` and `brand`.

### Solution

```python
class Car:
    wheels = 4   # class attribute

    def __init__(self, brand):
        self.brand = brand   # instance attribute

car1 = Car("Toyota")
car2 = Car("Honda")

print(car1.brand, car1.wheels)
print(car2.brand, car2.wheels)
```
