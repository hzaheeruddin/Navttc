## Task 11: A Simple `Dataset` Class
Create a class `Dataset` that stores a list of numeric values (passed into `__init__`). Add methods:
- `mean()` — returns the average of the values.
- `minimum()` — returns the smallest value.
- `maximum()` — returns the largest value.
- `normalize()` — returns a **new list** where each value is scaled to the range `[0, 1]` using `(value - minimum) / (maximum - minimum)`.

Test all four methods on `Dataset([10, 20, 30, 40, 50])`.

### Solution

```python
class Dataset:
    def __init__(self, values):
        self.values = values

    def mean(self):
        return sum(self.values) / len(self.values)

    def minimum(self):
        return min(self.values)

    def maximum(self):
        return max(self.values)

    def normalize(self):
        low = self.minimum()
        high = self.maximum()
        return [(v - low) / (high - low) for v in self.values]


data = Dataset([10, 20, 30, 40, 50])
print("Mean:", data.mean())
print("Min:", data.minimum())
print("Max:", data.maximum())
print("Normalized:", data.normalize())
```
