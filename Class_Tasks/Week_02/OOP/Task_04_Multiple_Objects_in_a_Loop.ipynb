## Task 4: Multiple Objects in a Loop
Create a class `Employee` with `name` and `salary` attributes. Create a list of at least 3 `Employee` objects. Loop through the list and print each employee's name and salary using an f-string.

### Solution

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

employees = [
    Employee("Ali", 50000),
    Employee("Sara", 65000),
    Employee("Ahmed", 48000),
]

for emp in employees:
    print(f"{emp.name}: {emp.salary}")
```
