## Task 13: Records as Objects
You previously worked with student records as dictionaries: `{"name": "Ali", "marks": [78, 85, 90]}`. Rewrite this idea using a class `StudentRecord` with `name` and `marks` attributes, and methods `average()` and `grade()` (same grading rule as earlier notebooks: `>=90` → `"A"`, `>=75` → `"B"`, `>=60` → `"C"`, else `"F"`). Create a list of at least 4 `StudentRecord` objects, then use a list comprehension to build a list of names of students with grade `"A"` or `"B"`.

### Solution

```python
class StudentRecord:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()
        if avg >= 90:
            return "A"
        elif avg >= 75:
            return "B"
        elif avg >= 60:
            return "C"
        else:
            return "F"


students = [
    StudentRecord("Ali", [78, 85, 90]),
    StudentRecord("Sara", [88, 92, 95]),
    StudentRecord("Ahmed", [55, 62, 58]),
    StudentRecord("Ayesha", [90, 87, 93]),
]

top_students = [s.name for s in students if s.grade() in ("A", "B")]
print(top_students)
```
