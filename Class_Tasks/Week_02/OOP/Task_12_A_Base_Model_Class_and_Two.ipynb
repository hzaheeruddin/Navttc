## Task 12: A Base `Model` Class and Two Subclasses
Create a base class `BaseModel` with an `__init__` that sets `self.is_trained = False`, and a method `predict()` that raises a `ValueError` if `self.is_trained` is `False` (mirroring the `SimpleAverageModel` example from the concepts notebook).

Create two child classes:
- `ConstantModel` — its `fit(training_values)` stores a fixed constant (e.g. the maximum of `training_values`) and sets `self.is_trained = True`; its `predict(n)` returns a list of `n` copies of that constant.
- `AverageModel` — its `fit(training_values)` stores the average, and `predict(n)` returns `n` copies of the average.

Both should reuse the parent's `predict()` guard logic somehow (hint: you can still call `super().predict()` first inside the child's `predict()`, purely to trigger the "not trained" check, before returning your own result — or re-check `self.is_trained` directly).

Train both models on `[5, 10, 15, 20, 25]`, call `predict(3)` on each, and print the results. Then create a third, **untrained** model of either type and confirm calling `predict()` raises the expected error.

### Solution

```python
class BaseModel:
    def __init__(self):
        self.is_trained = False

    def predict(self):
        if not self.is_trained:
            raise ValueError("Model has not been trained yet. Call fit() first.")


class ConstantModel(BaseModel):
    def __init__(self):
        super().__init__()
        self.constant = None

    def fit(self, training_values):
        self.constant = max(training_values)
        self.is_trained = True

    def predict(self, n):
        super().predict()   # reuse the parent's "not trained" guard
        return [self.constant] * n


class AverageModel(BaseModel):
    def __init__(self):
        super().__init__()
        self.average = None

    def fit(self, training_values):
        self.average = sum(training_values) / len(training_values)
        self.is_trained = True

    def predict(self, n):
        super().predict()
        return [self.average] * n


training_data = [5, 10, 15, 20, 25]

constant_model = ConstantModel()
constant_model.fit(training_data)
print(constant_model.predict(3))

average_model = AverageModel()
average_model.fit(training_data)
print(average_model.predict(3))

untrained_model = AverageModel()
try:
    untrained_model.predict(2)
except ValueError as error:
    print("Error:", error)
```
