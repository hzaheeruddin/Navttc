## Task 10: Combine Class + List Comprehension
Create a class `Product` with `name` and `price` attributes. Create a list of at least 4 `Product` objects with different prices. Using a list comprehension, build a new list containing only the **names** of products priced above `100`.

### Solution

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

products = [
    Product("Keyboard", 45),
    Product("Monitor", 150),
    Product("Mouse", 25),
    Product("Laptop", 1200),
]

expensive_names = [p.name for p in products if p.price > 100]
print(expensive_names)
```
