## Task 9: Combining Arrays
Create two 1D arrays, `group_a = np.array([1, 2, 3])` and `group_b = np.array([4, 5, 6])`. Use `np.concatenate` to join them, and separately use `np.vstack` to stack them as two rows of a 2D array. Print both results.

### Setup

```python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
```

### Solution

```python
group_a = np.array([1, 2, 3])
group_b = np.array([4, 5, 6])

print(np.concatenate([group_a, group_b]))
print(np.vstack([group_a, group_b]))
```
