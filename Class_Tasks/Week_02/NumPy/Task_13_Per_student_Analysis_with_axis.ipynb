## Task 13: Per-student Analysis with `axis`
You are given marks for 4 students across 3 subjects:
```python
student_marks = np.array([
    [78, 85, 90],
    [88, 92, 95],
    [55, 62, 58],
    [90, 87, 93],
])
```
Using only vectorized NumPy operations (no Python loops):
1. Compute each student's average mark (`axis=1`).
2. Compute the class average for each subject (`axis=0`).
3. Use boolean indexing to find which students have an average `>= 80`.
4. Save the array of per-student averages to a file `student_averages.npy`, then load it back and print it.

### Setup

````python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
````

### Solution

````python
student_marks = np.array([
    [78, 85, 90],
    [88, 92, 95],
    [55, 62, 58],
    [90, 87, 93],
])

student_averages = student_marks.mean(axis=1)
subject_averages = student_marks.mean(axis=0)
top_students_mask = student_averages >= 80

print("Per-student averages:", student_averages)
print("Per-subject averages:", subject_averages)
print("Students with average >= 80:", student_averages[top_students_mask])

np.save("sample_data/student_averages.npy", student_averages)
loaded_averages = np.load("sample_data/student_averages.npy")
print("Loaded back from disk:", loaded_averages)
````
