## Task 6: Filtering with Boolean Indexing
Given `scores = np.array([55, 89, 76, 92, 40, 67, 98])`, extract only the scores that are `>= 70` using boolean indexing. Then print how many scores passed that filter.

### Setup

```python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
```

### Solution

```python
scores = np.array([55, 89, 76, 92, 40, 67, 98])

passing_scores = scores[scores >= 70]
print(passing_scores)
print("Number passing:", passing_scores.size)
```
