## Task 5: Element-wise Arithmetic
Create `celsius = np.array([0, 10, 20, 30, 40])`. Convert it to Fahrenheit using the formula `F = C * 9/5 + 32`, in a single vectorized line (no loop).

### Setup

```python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
```

### Solution

```python
celsius = np.array([0, 10, 20, 30, 40])
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit)
```
