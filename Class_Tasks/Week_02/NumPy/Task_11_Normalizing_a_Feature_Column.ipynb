## Task 11: Normalizing a Feature Column
You are given a raw feature column representing house sizes in square feet:
```python
house_sizes = np.array([850, 900, 1200, 1500, 2000, 750, 1100])
```
Normalize this array so every value is scaled into the range `[0, 1]`, using the formula `(value - min) / (max - min)`. Do this in a single vectorized line (no loop), then print the result.

### Setup

````python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
````

### Solution

````python
house_sizes = np.array([850, 900, 1200, 1500, 2000, 750, 1100])

normalized = (house_sizes - house_sizes.min()) / (house_sizes.max() - house_sizes.min())
print(normalized)
````
