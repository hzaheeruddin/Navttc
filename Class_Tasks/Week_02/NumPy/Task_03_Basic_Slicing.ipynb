## Task 3: Basic Slicing
Create the array `values = np.arange(10, 60, 5)`. Print the first 3 elements, the last 3 elements, and every second element.

### Setup

```python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
```

### Solution

```python
values = np.arange(10, 60, 5)

print(values[:3])
print(values[-3:])
print(values[::2])
```
