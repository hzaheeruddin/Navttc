## Task 7: Reshape and Aggregate
Create `numbers = np.arange(1, 25)`. Reshape it into a `(4, 6)` array. Print the sum of each row (`axis=1`) and the sum of each column (`axis=0`).

### Setup

```python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
```

### Solution

```python
numbers = np.arange(1, 25)
grid = numbers.reshape(4, 6)

print("Row sums:", grid.sum(axis=1))
print("Column sums:", grid.sum(axis=0))
```
