## Task 2: Generator Functions
Create a `4x4` array of zeros, a `3x3` array filled with the value `9`, and a `5x5` identity matrix. Print all three.

### Setup

```python
import numpy as np
import os
os.makedirs("sample_data", exist_ok=True)   # folder used by the save/load task
```

### Solution

```python
zeros_4x4 = np.zeros((4, 4))
nines_3x3 = np.full((3, 3), 9)
identity_5x5 = np.eye(5)

print(zeros_4x4)
print(nines_3x3)
print(identity_5x5)
```
