## Task 4: Row Selection with `.iloc[]`
Print the 4th row of `df` (by position, not label) using `.iloc[]`. Then print just the `city` value of that same row.

### Prerequisites
Runs the setup and the earlier steps this task depends on (output hidden).

```python
%%capture
import pandas as pd
import numpy as np
import os

os.makedirs("sample_data", exist_ok=True)

sample_csv_text = """name,age,city,department,salary
Ali,28,Lahore,Engineering,85000
Sara,32,Karachi,Marketing,72000
Ahmed,45,Lahore,Engineering,120000
Ayesha,24,Islamabad,Sales,55000
Bilal,38,Karachi,Engineering,98000
Zainab,29,Islamabad,Marketing,68000
Usman,52,Lahore,Sales,60000
Hina,31,Karachi,Engineering,91000
"""

with open("sample_data/employees.csv", "w") as file:
    file.write(sample_csv_text)

print("employees.csv is ready.")
print(pd.__version__)

df = pd.read_csv("sample_data/employees.csv")
print(df.head(3))
print("Shape:", df.shape)
print(df.dtypes)

department_series = df["department"]
print(department_series.unique())
print("Unique departments:", department_series.nunique())

name_salary = df[["name", "salary"]]
print(name_salary)
```

### Solution

```python
print(df.iloc[3])                      # 4th row by position
print(df.iloc[3]["city"])              # just its city value
```
