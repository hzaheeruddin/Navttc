## Task 3 - Bonus: Type Casting
Pick **one** customer's `monthly_charges` and cast it to `int`.
1. Print the original float value.
2. Print the casted int value.
3. In a comment, explain what value was lost and why (hint: `int()` truncates, it does not round).

### Prerequisites
Runs the setup and the earlier steps this task depends on (output hidden).

```python
%%capture
# Customer 1
customer1_tenure = 5
customer1_charges = 29.85

# Customer 2
customer2_tenure = 34
customer2_charges = 56.95

# Customer 3
customer3_tenure = 24
customer3_charges = 89.10

for n, (t, c) in enumerate([(customer1_tenure, customer1_charges),
                            (customer2_tenure, customer2_charges),
                            (customer3_tenure, customer3_charges)], start=1):
    print(f"Customer {n}: tenure={t} ({type(t).__name__}), charges={c} ({type(c).__name__})")

# Customer 1 check
c1_flag = customer1_tenure > 12 and customer1_charges > 70
print("Customer 1 is high value & long-term:", c1_flag)

# Customer 2 check
c2_flag = customer2_tenure > 12 and customer2_charges > 70
print("Customer 2 is high value & long-term:", c2_flag)

# Customer 3 check
c3_flag = customer3_tenure > 12 and customer3_charges > 70
print("Customer 3 is high value & long-term:", c3_flag)
```

### Solution

```python
print("Original (float):", customer3_charges)
casted = int(customer3_charges)
print("Casted (int):   ", casted)
# The fractional part .10 was lost: int() truncates toward zero instead of rounding,
# so 89.10 became 89. (A value like 89.99 would also become 89, not 90.)
```
