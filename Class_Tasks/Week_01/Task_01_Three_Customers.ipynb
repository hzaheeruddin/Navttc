## Task 1 - Three Customers
Create variables for **three fictional customers**, each with:
- `tenure_months` (an `int`)
- `monthly_charges` (a `float`)

Name them clearly, e.g. `customer1_tenure`, `customer1_charges`, etc.

### Solution

```python
# Customer 1
customer1_tenure = 5
customer1_charges = 29.85

# Customer 2
customer2_tenure = 34
customer2_charges = 56.95

# Customer 3
customer3_tenure = 24
customer3_charges = 89.10

for n, (t, c) in enumerate([(customer1_tenure, customer1_charges),
                            (customer2_tenure, customer2_charges),
                            (customer3_tenure, customer3_charges)], start=1):
    print(f"Customer {n}: tenure={t} ({type(t).__name__}), charges={c} ({type(c).__name__})")
```
