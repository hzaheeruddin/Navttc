### Dimensions: 0-D, 1-D, 2-D, 3-D

```python
a0 = np.array(42)
a1 = np.array([1, 2, 3])
a2 = np.array([[1, 2, 3], [4, 5, 6]])
a3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
for a in (a0, a1, a2, a3): print(a.ndim, a.shape)
```

```python
print(np.array([1, 2, 3, 4], ndmin=5))
```

### Built-in constructors

```python
print(np.zeros((2, 3)))
print(np.ones((2, 2), dtype=int))
print(np.full((2, 2), 7))
print(np.eye(3))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.empty((2, 2)).shape)
```

```python
print(np.array(list(range(5))))   # from python list
print(np.random.rand(2, 3))         # from random
```
