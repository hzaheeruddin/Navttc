**Description:** Exercise on Numpy Array splitting.

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5, 6])
print(np.array_split(arr, 3))
print(np.array_split(arr, 4))   # uneven allowed
```

```python
parts = np.array_split(arr, 3); print(parts[0], parts[1], parts[2])
```

```python
a2 = np.arange(1, 13).reshape(4, 3)
print(np.array_split(a2, 2))
print(np.array_split(a2, 3, axis=1))
print(np.hsplit(a2, 3))
print(np.vsplit(a2, 2))
```

```python
print(np.split(arr, [2, 4]))   # split at indices
```
