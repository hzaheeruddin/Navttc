# Task 32: NumPy - Copy vs View

**Week:** 3  
**Description:** Exercise on Numpy Array copying and viewing.

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
x = arr.copy(); arr[0] = 42
print('original:', arr, '| copy:', x)
```

```python
arr = np.array([1, 2, 3, 4, 5])
y = arr.view(); arr[0] = 42
print('original:', arr, '| view:', y)
```

```python
y[1] = 99          # changes in view also change the original
print(arr, y)
```

```python
# .base tells you who owns the data
c = arr.copy(); v = arr.view()
print('copy.base:', c.base, '| view.base is arr:', v.base is arr)
```

```python
print(np.shares_memory(arr, v), np.shares_memory(arr, c))
```
