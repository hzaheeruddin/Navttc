**Week:** 3  
**Description:** Exercise on Numpy Array iteration.

```python
import numpy as np
for x in np.array([1, 2, 3]): print(x, end=' ')
```

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
for row in a: print(row)
for row in a:
    for x in row: print(x, end=' ')
```

### np.nditer, ndenumerate

```python
for x in np.nditer(a): print(x, end=' ')
print()
for idx, x in np.ndenumerate(a): print(idx, x)
```

```python
for x in np.nditer(a, flags=['buffered'], op_dtypes=['S']): print(x, end=' ')
print()
for x in np.nditer(a[:, ::2]): print(x, end=' ')
```

Prefer vectorized ops over loops:

```python
big = np.arange(1_000_000)
%timeit big * 2
%timeit [x * 2 for x in range(1_000_000)]
```
