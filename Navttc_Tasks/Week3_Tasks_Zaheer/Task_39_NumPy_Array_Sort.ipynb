```python
import numpy as np
print(np.sort(np.array([3, 2, 0, 1])))
print(np.sort(np.array(['banana', 'cherry', 'apple'])))
print(np.sort(np.array([True, False, True])))
```

```python
a2 = np.array([[3, 2, 4], [5, 0, 1]])
print(np.sort(a2))             # sorts each row
print(np.sort(a2, axis=0))
print(np.sort(a2)[:, ::-1])    # descending
```

```python
x = np.array([30, 10, 20])
print(np.argsort(x), x[np.argsort(x)])
```

```python
# sort one array by another
names = np.array(['A', 'B', 'C']); score = np.array([70, 95, 80])
print(names[np.argsort(-score)])
print(np.lexsort((names, score)))
```
