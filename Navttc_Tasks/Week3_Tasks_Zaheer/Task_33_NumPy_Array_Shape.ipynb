```python
import numpy as np
a = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print('shape:', a.shape, '| ndim:', a.ndim, '| size:', a.size, '| dtype:', a.dtype, '| itemsize:', a.itemsize)
```

```python
b = np.array([1, 2, 3, 4], ndmin=5)
print(b.shape)   # 5 dimensions, last has 4 elements
```

```python
print(np.zeros((3, 4, 2)).shape)
print(a.T.shape)
```
