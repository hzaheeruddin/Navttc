```python
import numpy as np
from numpy import random
print(random.randint(100), random.rand(), random.rand(2, 3), random.randint(100, size=(2, 3)))
```

```python
print(random.choice([3, 5, 7, 9]), random.choice([3, 5, 7, 9], size=(2, 3)))
print(random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=10))
```

```python
rng = np.random.default_rng(42)         # modern, reproducible API
print(rng.integers(0, 10, 5), rng.random(3), rng.normal(0, 1, 3))
```

```python
arr = np.arange(1, 8)
rng.shuffle(arr); print(arr)
print(rng.permutation(np.arange(1, 8)))
```

### Distributions

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 3, figsize=(12, 3))
ax[0].hist(rng.normal(50, 10, 5000), bins=30); ax[0].set_title('normal')
ax[1].hist(rng.uniform(0, 1, 5000), bins=30); ax[1].set_title('uniform')
ax[2].hist(rng.binomial(10, 0.5, 5000), bins=11); ax[2].set_title('binomial')
plt.show()
```
