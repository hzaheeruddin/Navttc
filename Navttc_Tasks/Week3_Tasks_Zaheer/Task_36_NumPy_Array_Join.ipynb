# Task 36: NumPy - Array Join

**Week:** 3  
**Description:** Exercise on Numpy Matrix joining.

```python
import numpy as np
a, b = np.array([1, 2, 3]), np.array([4, 5, 6])
print(np.concatenate((a, b)))
```

```python
a2, b2 = np.array([[1, 2], [3, 4]]), np.array([[5, 6], [7, 8]])
print(np.concatenate((a2, b2), axis=0))
print(np.concatenate((a2, b2), axis=1))
```

```python
print(np.stack((a, b), axis=1))
print(np.hstack((a, b)))
print(np.vstack((a, b)))
print(np.dstack((a, b)))
```

```python
print(np.append(a, [7, 8]))
print(np.insert(a, 1, 99))
print(np.delete(a, 0))
```
