# Task 38: NumPy - Array Search

**Week:** 3  
**Description:** Exercise on Numpy Array searching.

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5, 4, 4])
print(np.where(arr == 4))
print(np.where(arr % 2 == 0))
```

```python
sorted_arr = np.array([6, 7, 8, 9])
print(np.searchsorted(sorted_arr, 7))
print(np.searchsorted(sorted_arr, 7, side='right'))
print(np.searchsorted(sorted_arr, [2, 4, 11]))
```

```python
print(np.argmax(arr), np.argmin(arr), arr.max(), arr.min())
print(np.isin(arr, [1, 5]), np.any(arr > 4), np.all(arr > 0))
```

```python
print(np.where(arr > 3, 'big', 'small'))
print(np.nonzero(arr > 3))
```
