```python
import numpy as np
print(np.array([1, 2, 3]).dtype, np.array([1.5, 2]).dtype, np.array(['a', 'bc']).dtype, np.array([True, False]).dtype)
```

```python
arr = np.array([1, 2, 3, 4], dtype='S')
print(arr, arr.dtype)
print(np.array([1, 2, 3, 4], dtype='i4').dtype, np.array([1, 2, 3], dtype='f8').dtype)
```

### Converting with astype()

```python
f = np.array([1.1, 2.1, 3.1])
print(f.astype('i'), f.astype(int).dtype)
print(np.array([1, 0, 3]).astype(bool))
```

### Type codes: `i` int, `u` uint, `f` float, `b` bool, `S` bytes, `U` unicode, `c` complex

```python
print(np.iinfo(np.int8), np.finfo(np.float32).eps)
```
