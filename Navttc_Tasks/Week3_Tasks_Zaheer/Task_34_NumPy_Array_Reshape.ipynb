**Description:** Exercise on Numpy Array reshaping.

```python
import numpy as np
arr = np.arange(1, 13)
print(arr.reshape(4, 3))
print(arr.reshape(2, 3, 2))
```

```python
print(arr.reshape(2, -1).shape, arr.reshape(-1, 6).shape)   # -1 = infer
```

```python
r = arr.reshape(3, 4)
print(r.base is arr)    # reshape returns a view when possible
```

### Flattening

```python
print(r.flatten(), r.ravel(), r.reshape(-1))
```

```python
print(arr.reshape(3, 4).T)        # transpose
print(np.expand_dims(arr, 0).shape, np.squeeze(np.expand_dims(arr, 0)).shape)
```
