```python
try:
    print(10 / 0)
except ZeroDivisionError as e:
    print('Error:', e)
```

```python
def parse(s):
    try:
        return int(s)
    except ValueError:
        print(f'{s!r} is not a number')
    except TypeError:
        print('wrong type')
    else:
        print('parsed without error')
    finally:
        print('finally always runs')
parse('42'); parse('abc'); parse(None)
```

```python
# multiple exceptions in one clause
try:
    [1, 2][5]
except (IndexError, KeyError) as e:
    print(type(e).__name__, e)
```

```python
# raise exceptions
def set_age(a):
    if a < 0: raise ValueError('age cannot be negative')
    return a
try: set_age(-1)
except ValueError as e: print(e)
```

```python
# file handling errors
try:
    open('missing_file.txt')
except FileNotFoundError as e:
    print('File not found:', e)
```

```python
# safe file reading with context manager
with open('demo.txt', 'w') as f: f.write('hi\nthere')
with open('demo.txt') as f: print(f.read())
import os; os.remove('demo.txt')
```

```python
# assert
try: assert 1 + 1 == 3, 'math is broken'
except AssertionError as e: print('AssertionError:', e)
```
