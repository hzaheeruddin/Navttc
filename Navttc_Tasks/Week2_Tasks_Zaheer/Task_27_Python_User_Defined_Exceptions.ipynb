**Description:** Exercise on User defined Exception Handling in Python.

```python
class Error(Exception):
    '''Base class for custom exceptions'''

class ValueTooSmallError(Error):
    pass

class ValueTooLargeError(Error):
    pass

def check(n, target=10):
    if n < target: raise ValueTooSmallError('Value is too small')
    if n > target: raise ValueTooLargeError('Value is too large')
    return 'Correct!'

for n in (5, 15, 10):
    try:
        print(check(n))
    except ValueTooSmallError as e: print(n, '->', e)
    except ValueTooLargeError as e: print(n, '->', e)
```

### Custom exception with extra data

```python
class InsufficientBalance(Exception):
    def __init__(self, balance, amount):
        self.balance, self.amount = balance, amount
        super().__init__(f'Balance {balance} is less than requested {amount}')

class Account:
    def __init__(self, balance): self.balance = balance
    def withdraw(self, amt):
        if amt > self.balance: raise InsufficientBalance(self.balance, amt)
        self.balance -= amt

acc = Account(100)
try:
    acc.withdraw(500)
except InsufficientBalance as e:
    print(e, '| short by', e.amount - e.balance)
```

```python
# exception chaining
try:
    try: int('x')
    except ValueError as e: raise Error('conversion failed') from e
except Error as e:
    print(e, '| caused by:', repr(e.__cause__))
```
