**Description:** Perform Standard Deviation.

```python
import numpy as np
speed = [86, 87, 88, 86, 87, 85, 86]
print('std:', np.std(speed), '| var:', np.var(speed))
```

```python
speed2 = [32, 111, 138, 28, 59, 77, 97]
print('std:', np.std(speed2), '| var:', np.var(speed2))
print('Higher std -> values are more spread out')
```

### By hand: population vs sample

```python
x = np.array(speed2)
mean = x.mean()
var_pop = ((x - mean) ** 2).sum() / len(x)
var_samp = ((x - mean) ** 2).sum() / (len(x) - 1)
print('population std:', var_pop ** 0.5, np.std(x))
print('sample std    :', var_samp ** 0.5, np.std(x, ddof=1))
```

### Coefficient of variation, z-scores, 68-95-99.7 rule

```python
print('CV %:', np.std(x) / x.mean() * 100)
print('z-scores:', np.round((x - x.mean()) / x.std(), 2))
data = np.random.default_rng(0).normal(100, 15, 100000)
for k in (1, 2, 3):
    print(f'within {k} std: {np.mean(abs(data-100) < k*15):.3f}')
```
