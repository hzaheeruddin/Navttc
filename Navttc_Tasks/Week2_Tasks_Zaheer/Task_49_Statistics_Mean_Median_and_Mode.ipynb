# Task 49: Statistics - Mean, Median and Mode

**Week:** 2  
**Description:** Perform Mean, Median and Mode.

```python
import numpy as np
from scipy import stats
speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
```

```python
print('Mean  :', np.mean(speed))
print('Median:', np.median(speed))
print('Mode  :', stats.mode(speed, keepdims=False).mode)
```

### Same with the standard library `statistics` module

```python
import statistics as st
print(st.mean(speed), st.median(speed), st.mode(speed), st.multimode(speed))
```

### Effect of an outlier

```python
with_out = speed + [500]
print('mean  :', np.mean(speed), '->', round(np.mean(with_out), 2))
print('median:', np.median(speed), '->', np.median(with_out))
```

```python
import matplotlib.pyplot as plt
plt.hist(speed, bins=8, edgecolor='k')
plt.axvline(np.mean(speed), color='r', label='mean')
plt.axvline(np.median(speed), color='g', ls='--', label='median')
plt.legend(); plt.title('Speed distribution'); plt.show()
```
