**Description:** Exercise on User defined functions in Python.

```python
def greet(name):
    return f'Hello, {name}!'
print(greet('Ali'))
```

```python
def power(base, exp=2):   # default argument
    return base ** exp
print(power(3), power(2, 5), power(exp=3, base=2))
```

```python
def total(*args, **kwargs):  # variable args
    print('args:', args, 'kwargs:', kwargs)
    return sum(args)
print(total(1, 2, 3, scale=10))
```

```python
def min_max(nums):          # multiple return values
    return min(nums), max(nums)
lo, hi = min_max([4, 1, 9]); print(lo, hi)
```

```python
# scope: local vs global
x = 'global'
def f():
    x = 'local'
    return x
print(f(), x)
count = 0
def inc():
    global count
    count += 1
inc(); inc(); print(count)
```

```python
def fact(n):                # recursion
    return 1 if n <= 1 else n * fact(n - 1)
print(fact(6))
```

```python
# lambda, map, filter, nested functions
sq = lambda x: x * x
print(list(map(sq, [1, 2, 3])), list(filter(lambda x: x % 2, range(10))))
def outer(a):
    def inner(b): return a + b
    return inner
print(outer(3)(4))
```

```python
def add(a: int, b: int = 0) -> int:
    '''Add two integers.'''
    return a + b
print(add.__doc__, add(2, 3))
```
