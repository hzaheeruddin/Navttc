# Task 2: Linux Commands

**Week:** 1  
**Description:** Practice pwd, cd, ls, cat, sudo, man, redirection, mkdir, rm, rmdir, cp, mv, file, more, less, head, alias, shutdown, restart, touch, nano, bash, sh, chmod, ps, kill, dpkg.

Run this notebook on Linux/macOS/WSL. Cells use `%%bash` and work inside a safe sandbox folder (`~/linux_practice`).
Cells marked **(manual)** need an interactive terminal.

```python
%%bash
mkdir -p ~/linux_practice && cd ~/linux_practice && pwd   # pwd + mkdir + cd
```

```python
%%bash
cd ~/linux_practice
touch a.txt b.txt            # touch
ls                            # ls
ls -l                         # long listing
ls -a                         # hidden files too
```

```python
%%bash
cd ~/linux_practice
echo 'Hello Linux' > a.txt        # redirection (overwrite)
echo 'Second line' >> a.txt         # redirection (append)
cat a.txt                           # cat
```

```python
%%bash
cd ~/linux_practice
seq 1 50 > numbers.txt
head numbers.txt            # first 10 lines
head -n 3 numbers.txt
tail -n 3 numbers.txt
file numbers.txt            # file type
# more / less are interactive pagers (manual): more numbers.txt ; less numbers.txt
```

```python
%%bash
cd ~/linux_practice
mkdir -p dir1/sub
cp a.txt dir1/a_copy.txt       # cp
mv b.txt dir1/sub/b_moved.txt # mv (move)
mv dir1/a_copy.txt dir1/renamed.txt   # mv (rename)
ls -R
```

```python
%%bash
cd ~/linux_practice
rm a.txt                  # remove file
rm -r dir1/sub            # remove directory recursively
rmdir dir1 2>&1 || echo 'rmdir only removes EMPTY directories'
ls -R
```

```python
%%bash
cd ~/linux_practice
touch script.sh
echo '#!/bin/bash' > script.sh
echo 'echo Script is running as $(whoami)' >> script.sh
ls -l script.sh
chmod +x script.sh        # chmod
ls -l script.sh
./script.sh               # run directly
bash script.sh            # bash
sh script.sh              # sh
```

```python
%%bash
man ls | head -n 15   # man (manual pages)
```

```python
%%bash
alias ll='ls -la'      # alias (lasts only in this shell; put in ~/.bashrc to persist)
alias ll
shopt -s expand_aliases
ll ~/linux_practice | head
```

```python
%%bash
ps                       # processes in this shell
ps aux | head -n 5
sleep 300 &               # start a background process
PID=$!
echo "Started sleep with PID $PID"
kill $PID                 # kill it
sleep 1; ps -p $PID || echo 'process ended'
```

```python
%%bash
dpkg --version | head -n 1      # Debian package manager
dpkg -l | head -n 5           # list installed packages
dpkg -s bash | head -n 5      # package status
# sudo apt update && sudo apt upgrade   (manual - needs sudo)
```

```python
%%bash
sudo -n true 2>/dev/null && echo 'sudo available' || echo 'sudo needs password / not available here (manual)'
```
