# Task 1: Installation

**Week:** 1  
**Description:** Download and install Anaconda3, PyTorch, TensorFlow 2.0, VSCode, PyCharm.

## Step-by-step
1. **Anaconda3** - https://www.anaconda.com/download (choose your OS, run installer, tick "Add to PATH" on Windows only if you know why).
2. **VSCode** - https://code.visualstudio.com/ (install the *Python* and *Jupyter* extensions).
3. **PyCharm** - https://www.jetbrains.com/pycharm/download/ (Community edition is free).
4. Create a conda environment and install the libraries (run in Anaconda Prompt / terminal):

```bash
conda create -n ai_course python=3.10 -y
conda activate ai_course
pip install numpy pandas matplotlib seaborn scikit-learn jupyter nltk
pip install tensorflow          # TensorFlow 2.x
pip install torch torchvision   # PyTorch (see https://pytorch.org for the exact command for your OS/GPU)
```

## Verify the installation

````python
import sys, importlib
print("Python:", sys.version)
for pkg in ["numpy", "pandas", "matplotlib", "seaborn", "sklearn", "nltk", "tensorflow", "torch"]:
    try:
        m = importlib.import_module(pkg)
        print(f"{pkg:12s} OK   version {getattr(m, '__version__', '?')}")
    except ImportError:
        print(f"{pkg:12s} NOT INSTALLED")
````

````python
# TensorFlow / PyTorch quick smoke tests (skipped automatically if not installed)
try:
    import tensorflow as tf
    print("TF:", tf.__version__, "| 2+3 =", tf.add(2, 3).numpy())
except ImportError:
    print("TensorFlow not installed")
try:
    import torch
    print("Torch:", torch.__version__, "| CUDA available:", torch.cuda.is_available())
except ImportError:
    print("PyTorch not installed")
````
