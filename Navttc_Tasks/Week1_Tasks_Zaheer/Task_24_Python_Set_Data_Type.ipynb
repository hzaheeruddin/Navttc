**Week:** 1  
**Description:** Exercise on Set data type.

```python
s = {1, 2, 3, 3, 2}
print(s, type(s), set([1, 1, 2]), set())
```

```python
s.add(4); s.update([5, 6]); s.discard(10); s.remove(1)
print(s)
```

```python
a, b = {1, 2, 3, 4}, {3, 4, 5, 6}
print('union', a | b, a.union(b))
print('intersection', a & b)
print('difference', a - b)
print('symmetric diff', a ^ b)
```

```python
print({1, 2} <= {1, 2, 3}, {1, 2, 3} >= {1}, {1, 2}.isdisjoint({3}))
```

```python
# remove duplicates while counting uniques
data = [1, 2, 2, 3, 3, 3]
print(len(set(data)), sorted(set(data)))
```

```python
fs = frozenset([1, 2, 3])   # immutable set
print(fs)
```
