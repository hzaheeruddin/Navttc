# Task 3: Python - Add Two Numbers

**Week:** 1  
**Description:** Add two numbers (given code).

```python
# This program adds two numbers

num1 = 1.5
num2 = 6.3

# Add two numbers
sum = num1 + num2

# Display the sum
print('The sum of {0} and {1} is {2}'.format(num1, num2, sum))
```

Note: naming a variable `sum` shadows the built-in `sum()` function. Restore it:

```python
del sum
print(sum([1, 2, 3]))
```
