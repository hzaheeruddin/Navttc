```python
lst = [10, 20, 30, 40, 50]
print(lst[0], lst[-1], lst[1:4], lst[::-1])
```

```python
lst.append(60); lst.insert(1, 15); lst.extend([70, 80])
print(lst)
```

```python
lst.remove(15); popped = lst.pop(); del lst[0]
print(lst, popped)
```

```python
nums = [5, 2, 9, 1]
nums.sort(); print(nums)
nums.sort(reverse=True); print(nums)
print(sorted(nums), len(nums), max(nums), min(nums), sum(nums))
```

```python
a = [1, 2, 3]; b = a; c = a.copy()
a.append(4)
print(a, b, c)  # b is the same object, c is independent
```

```python
print([x * 2 for x in range(5) if x % 2 == 0])
nested = [[1, 2], [3, 4]]
print(nested[1][0], [n for row in nested for n in row])
```
