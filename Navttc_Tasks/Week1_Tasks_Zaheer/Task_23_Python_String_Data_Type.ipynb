```python
s = 'Hello, Python World'
print(s[0], s[-5:], s[::-1], len(s))
```

```python
print(s.upper(), s.lower(), s.title(), s.swapcase())
```

```python
print(s.split(', '), '-'.join(['a', 'b', 'c']), s.replace('Python', 'AI'))
```

```python
print(s.find('Python'), s.count('o'), s.startswith('Hello'), s.endswith('d'))
```

```python
print('  padded  '.strip(), 'abc'.center(9, '*'), '42'.zfill(5), 'abc'.isalpha(), '123'.isdigit())
```

```python
name, age = 'Sara', 22
print('Name: %s, Age: %d' % (name, age))
print('Name: {}, Age: {}'.format(name, age))
print(f'Name: {name}, Age: {age:>4}')
```

```python
# palindrome check
w = 'Level'.lower()
print(w == w[::-1])
```

```python
# strings are immutable
try:
    s[0] = 'J'
except TypeError as e:
    print(e)
```
