# Task 18: Python - Continue Statement

**Week:** 1  
**Description:** Exercise on continue statement.

```python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=' ')
```

### Skip a specific letter

```python
for ch in 'geeksforgeeks':
    if ch == 'e':
        continue
    print(ch, end='')
```

### continue in a while loop (must update counter first!)

```python
i = 0
while i < 8:
    i += 1
    if i % 3 == 0:
        continue
    print(i, end=' ')
```

### Practice: sum only positive numbers

```python
nums = [3, -1, 4, -5, 9]
total = 0
for n in nums:
    if n < 0: continue
    total += n
print(total)
```
