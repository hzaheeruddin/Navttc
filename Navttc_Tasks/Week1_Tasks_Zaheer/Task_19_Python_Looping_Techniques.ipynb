**Week:** 1  
**Description:** Exercise on various looping techniques.

```python
# enumerate
for i, v in enumerate(['tic', 'tac', 'toe']): print(i, v)
```

```python
# zip multiple sequences
q = ['name', 'quest', 'color']; a = ['Ali', 'AI', 'blue']
for x, y in zip(q, a): print(f'What is your {x}? {y}')
```

```python
# reversed and sorted
for i in reversed(range(1, 6)): print(i, end=' ')
print()
for f in sorted({'pear', 'apple', 'fig'}): print(f)
```

```python
# dict items()
knights = {'gallahad': 'the pure', 'robin': 'the brave'}
for k, v in knights.items(): print(k, v)
```

```python
# list comprehension, dict comprehension, generator expression
print([x**2 for x in range(6)])
print({x: x**2 for x in range(4)})
print(sum(x for x in range(10)))
```

```python
# iterating while modifying: iterate over a copy / build a new list
words = ['cat', 'window', 'defenestrate']
for w in words[:]:
    if len(w) > 6: words.insert(0, w)
print(words)
```

```python
# itertools
import itertools
print(list(itertools.chain([1, 2], [3, 4])))
print(list(itertools.product('AB', [1, 2])))
print(list(itertools.permutations([1, 2, 3], 2)))
```
