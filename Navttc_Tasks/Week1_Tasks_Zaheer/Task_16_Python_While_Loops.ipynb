### while-else

```python
n = 0
while n < 3:
    n += 1
else:
    print('while finished normally')
```

### Practice: countdown, digit sum, factorial, guess-until

```python
n = 5
while n > 0:
    print(n, end=' ')
    n -= 1
print('Liftoff!')

num, total = 12345, 0
while num:
    total += num % 10
    num //= 10
print('Digit sum:', total)

f, i = 1, 1
while i <= 6:
    f *= i; i += 1
print('6! =', f)
```

### Infinite loop with a break condition

```python
i = 0
while True:
    i += 1
    if i * i > 50:
        break
print('first i with i^2 > 50 is', i)
```
