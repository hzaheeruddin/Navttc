# Task 22: Python - Tuple Data Type

**Week:** 1  
**Description:** Exercise on Tuple data type.

```python
t = (1, 2, 3, 'a')
print(t[0], t[-1], t[1:3], len(t))
```

```python
single = (5,)    # trailing comma needed
print(type(single), type((5)))
```

```python
a, b, *rest = (1, 2, 3, 4, 5)
print(a, b, rest)
x, y = 1, 2; x, y = y, x; print(x, y)
```

```python
try:
    t[0] = 99
except TypeError as e:
    print('Tuples are immutable:', e)
```

```python
print((1, 2) + (3,), (0,) * 3, t.count(1), t.index('a'))
```

```python
# Tuples are hashable -> usable as dict keys
d = {(0, 0): 'origin', (1, 2): 'pt'}
print(d[(1, 2)])
```

```python
# Convert between list and tuple
print(tuple([1, 2]), list((3, 4)))
```
