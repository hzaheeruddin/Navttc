### 1. Looping over a sequence

```python
for x in ['apple', 'banana', 'cherry']:
    print(x)
```

### 2. range()

```python
for i in range(5): print(i, end=' ')
print()
for i in range(2, 10, 3): print(i, end=' ')
```

### 3. Loop over a string, tuple, dict

```python
for ch in 'Python': print(ch, end='-')
print()
d = {'a': 1, 'b': 2}
for k, v in d.items(): print(k, v)
```

### 4. enumerate and zip

```python
names = ['Ali', 'Sara', 'Omar']
scores = [90, 85, 70]
for i, n in enumerate(names, start=1): print(i, n)
for n, s in zip(names, scores): print(n, s)
```

### 5. for-else

```python
for i in range(3):
    print(i)
else:
    print('loop finished without break')
```
