### Find first prime above 100

```python
n = 101
while True:
    if all(n % d for d in range(2, int(n**0.5) + 1)):
        print('First prime > 100:', n)
        break
    n += 1
```

### break only exits the innermost loop

```python
for i in range(3):
    for j in range(3):
        if j == 1:
            break
        print(i, j)
```

### Search a list

```python
data = [4, 8, 15, 16, 23, 42]
for idx, v in enumerate(data):
    if v > 15:
        print('first value >15:', v, 'at index', idx)
        break
```
