```python
d = {'name': 'Ali', 'age': 21}
print(d['name'], d.get('city', 'N/A'))
```

```python
d['city'] = 'Karachi'; d['age'] = 22
del d['city']
print(d)
```

```python
print(d.keys(), d.values(), d.items())
for k, v in d.items(): print(k, '->', v)
```

```python
d.update({'course': 'AI', 'age': 23}); print(d.pop('course'), d.setdefault('lang', 'py'), d)
```

```python
sq = {x: x**2 for x in range(1, 6)}
print(sq)
print({v: k for k, v in sq.items()})
```

```python
students = {'a': {'marks': 80}, 'b': {'marks': 95}}
print(max(students, key=lambda k: students[k]['marks']))
```

```python
# word frequency
text = 'to be or not to be'
freq = {}
for w in text.split(): freq[w] = freq.get(w, 0) + 1
print(freq)
from collections import Counter; print(Counter(text.split()).most_common(2))
```
