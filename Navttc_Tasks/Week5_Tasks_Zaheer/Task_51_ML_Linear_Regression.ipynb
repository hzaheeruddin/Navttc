## 1. Simple (univariate) linear regression

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

rng = np.random.default_rng(0)
hours = rng.uniform(1, 10, 100)
score = 8 * hours + 20 + rng.normal(0, 6, 100)
df = pd.DataFrame({'Hours': hours, 'Score': score})
df.plot.scatter('Hours', 'Score'); plt.show()
```

```python
X, y = df[['Hours']], df['Score']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
model = LinearRegression().fit(X_train, y_train)
print('intercept:', model.intercept_, '| slope:', model.coef_[0])
```

```python
pred = model.predict(X_test)
print('MAE :', mean_absolute_error(y_test, pred))
print('RMSE:', mean_squared_error(y_test, pred) ** 0.5)
print('R2  :', r2_score(y_test, pred))
plt.scatter(X_test, y_test); plt.plot(X_test.sort_values('Hours'), model.predict(X_test.sort_values('Hours')), 'r'); plt.show()
```

## 2. Multiple linear regression (several features)

```python
n = 200
m = pd.DataFrame({'Area': rng.uniform(500, 3500, n), 'Rooms': rng.integers(1, 6, n), 'Age': rng.uniform(0, 40, n)})
m['Price'] = 120 * m.Area + 8000 * m.Rooms - 900 * m.Age + rng.normal(0, 15000, n)
Xm, ym = m.drop(columns='Price'), m['Price']
Xtr, Xte, ytr, yte = train_test_split(Xm, ym, test_size=.2, random_state=1)
lr = LinearRegression().fit(Xtr, ytr)
print(pd.DataFrame({'coef': lr.coef_}, index=Xm.columns))
print('R2:', r2_score(yte, lr.predict(Xte)))
```

```python
print('Predicted price for 2000 sqft, 3 rooms, 10 yrs:', lr.predict(pd.DataFrame([[2000, 3, 10]], columns=Xm.columns))[0])
```
