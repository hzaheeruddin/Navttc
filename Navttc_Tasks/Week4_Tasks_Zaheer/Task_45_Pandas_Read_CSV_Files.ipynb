We first create a sample `data.csv` so the notebook is self-contained.

```python
import pandas as pd
import numpy as np
rng = np.random.default_rng(1)
sample = pd.DataFrame({
    'Duration': rng.integers(30, 120, 20),
    'Pulse': rng.integers(90, 140, 20),
    'Maxpulse': rng.integers(120, 180, 20),
    'Calories': rng.normal(300, 60, 20).round(1)})
sample.to_csv('data.csv', index=False)
print('saved data.csv')
```

```python
df = pd.read_csv('data.csv')
print(df.head()); print(df.tail(3)); print(df.info())
```

```python
pd.set_option('display.max_rows', 5); print(df); pd.reset_option('display.max_rows')
print(pd.read_csv('data.csv', usecols=['Duration', 'Calories'], nrows=3))
```
