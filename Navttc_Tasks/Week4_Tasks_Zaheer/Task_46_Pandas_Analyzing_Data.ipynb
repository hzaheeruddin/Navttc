```python
import pandas as pd
import numpy as np
rng = np.random.default_rng(2)
df = pd.DataFrame({'Duration': rng.integers(30, 120, 50), 'Pulse': rng.integers(90, 140, 50), 'Calories': rng.normal(300, 60, 50).round(1), 'Type': rng.choice(['run', 'walk', 'bike'], 50)})
```

```python
print(df.head(3)); print(df.tail(3)); print(df.shape)
```

```python
df.info()
```

```python
print(df.describe())
print(df.describe(include='object'))
```

```python
print(df['Type'].value_counts()); print(df['Type'].unique(), df['Type'].nunique())
```

```python
print(df.groupby('Type')[['Duration', 'Calories']].mean())
```

```python
print(df.isnull().sum()); print(df.sort_values('Calories', ascending=False).head())
```
