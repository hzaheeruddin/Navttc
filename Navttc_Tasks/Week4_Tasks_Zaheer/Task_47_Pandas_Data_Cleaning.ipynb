**Description:** Exercise on Pandas Data Cleaning techniques.

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Date': ['2020/12/01', '2020/12/02', None, '20201226', '2020/12/04', '2020/12/04'],
    'Duration': [60, 60, 45, 45, 400, 400],
    'Calories': [409.1, 479.0, np.nan, 282.4, 406.0, 406.0]})
print(df)
```

### 1. Empty cells

```python
print(df.isnull().sum())
clean = df.dropna(subset=['Date'])
clean = clean.fillna({'Calories': clean['Calories'].mean()})
print(clean)
```

### 2. Wrong format

```python
clean['Date'] = pd.to_datetime(clean['Date'], format='mixed')
print(clean)
```

### 3. Wrong data (outliers)

```python
clean.loc[clean['Duration'] > 120, 'Duration'] = 120   # replace, or drop rows
print(clean)
```

### 4. Duplicates

```python
print(clean.duplicated())
clean = clean.drop_duplicates()
print(clean)
```
