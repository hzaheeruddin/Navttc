Pandas is a library for data analysis built on NumPy. Core objects: **Series** (1-D) and **DataFrame** (2-D).

```python
import pandas as pd
import numpy as np
print(pd.__version__)
df = pd.DataFrame({'cars': ['BMW', 'Volvo', 'Ford'], 'passings': [3, 7, 2]})
print(df)
```

```python
print(df.head(2), df.shape, df.dtypes, sep='\n\n')
```

```python
df['price'] = [50, 40, 30]
print(df[df['passings'] > 2])
print(df.describe())
```

```python
print(df.sort_values('price'), df['cars'].tolist(), df.mean(numeric_only=True), sep='\n\n')
```
