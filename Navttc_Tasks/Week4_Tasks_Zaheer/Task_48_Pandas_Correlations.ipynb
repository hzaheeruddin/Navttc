```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt, seaborn as sns
rng = np.random.default_rng(3)
dur = rng.integers(30, 120, 100)
df = pd.DataFrame({'Duration': dur,
    'Calories': dur * 3 + rng.normal(0, 30, 100),
    'Pulse': rng.integers(90, 140, 100)})
```

```python
print(df.corr())
```

Correlation near **1** = strong positive, near **-1** = strong negative, near **0** = no linear relationship. Correlation is not causation.

```python
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1); plt.show()
```

```python
df.plot.scatter('Duration', 'Calories', alpha=.6); plt.show()
print(df['Duration'].corr(df['Calories'], method='spearman'))
```
