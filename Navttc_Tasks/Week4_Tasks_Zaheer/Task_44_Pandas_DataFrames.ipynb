```python
import pandas as pd
import numpy as np
data = {'calories': [420, 380, 390], 'duration': [50, 40, 45]}
df = pd.DataFrame(data); print(df)
```

```python
print(df.loc[0]); print(df.loc[[0, 1]]); print(df.iloc[1, 0])
```

```python
df.index = ['day1', 'day2', 'day3']; print(df.loc['day2']); print(df['calories']); print(df[['calories', 'duration']])
```

```python
df['ratio'] = df['calories'] / df['duration']
df = df.rename(columns={'ratio': 'cal_per_min'})
print(df)
df = df.drop(columns='cal_per_min').drop(index='day1'); print(df)
```

```python
print(df.T); print(df.values); print(df.columns.tolist(), df.index.tolist())
print(df.apply(lambda col: col.max() - col.min()))
```
