```python
import pandas as pd
import numpy as np
s = pd.Series([1, 7, 2]); print(s); print(s[0])
```

```python
s = pd.Series([1, 7, 2], index=['x', 'y', 'z'], name='vals'); print(s['y'])
```

```python
s2 = pd.Series({'day1': 420, 'day2': 380, 'day3': 390}); print(s2)
print(pd.Series({'day1': 420, 'day2': 380, 'day3': 390}, index=['day1', 'day2']))
```

```python
print(s2.sum(), s2.mean(), s2.max(), s2.idxmax(), s2[s2 > 385], s2 * 2, sep='\n')
```

```python
print(s2.apply(lambda v: v / 10)); print(s2.to_dict()); print(s2.describe())
```
